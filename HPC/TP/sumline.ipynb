```python
import numpy as np
```

```python
def sum_cpu(mat):
    nraw, ncol = mat.shape
    res = np.zeros(nraw)
    for i in range(nraw):
        res[i] = np.sum(mat[i, :])
    return res
```

```python
M = np.random.randn(1024,1024)
M = M.astype(np.float32)
```

```python
res_c = sum_cpu(M)
```

```python
from numba import cuda
```

```python
@cuda.jit
def sum_gpu(mat,res):
    nraw, ncol = mat.shape
    i = cuda.threadIdx.x
    for j in range(ncol):
        res[i] += mat[i,j]
```

```python
nraw ,ncol = M.shape
# allocation on gpu
mat_gpu = cuda.to_device(M)
res_gpu = cuda.device_array((nraw))
```

```python
sum_gpu[1,nraw](mat_gpu, res_gpu)
```

```python
res = res_gpu.copy_to_host()
```
