```python
import numpy as np
from numba import cuda
```

```python
@cuda.jit
def doublify(a):
    tx = cuda.threadIdx.x 
    ty = cuda.threadIdx.y
    a[tx, ty] *= 2
```

```python
a = np.random.randn(4, 4)
```

```python
a_gpu = cuda.to_device(a)
```

```python
doublify[1,(4, 4)](a_gpu)
```

```python
a_doubled = a_gpu.copy_to_host()
```

```python
print(a_doubled)
print(a)
np.allclose(a * 2, a_doubled)
```
